```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
import os

# Set paths
input_folder = r"D:\CodAlpha\Task2\DATA"  # Root folder with subfolders containing .wav files
output_folder = r"D:\CodAlpha\Task2\MelSpectrograms"  # Root output folder to save PNGs

# Parameters for narrow-band spectrogram
n_fft = 512
hop_length = 512
window = 'hann'  # Hanning window

# Traverse all subdirectories
for root, dirs, files in os.walk(input_folder):
    for filename in files:
        if filename.endswith(".wav"):
            file_path = os.path.join(root, filename)
            
            # Load audio
            y, sr = librosa.load(file_path, sr=None)
            
            # Compute narrow-band Mel spectrogram
            S = librosa.feature.melspectrogram(
                y=y,
                sr=sr,
                n_fft=n_fft,
                hop_length=hop_length,
                window=window,
                n_mels=128,
                fmax=8000
            )
            S_dB = librosa.power_to_db(S, ref=np.max)

            # Plot without axes or labels
            fig, ax = plt.subplots(figsize=(10, 4))
            librosa.display.specshow(S_dB, sr=sr, hop_length=hop_length, fmax=8000, ax=ax)
            ax.axis('off')  # Hide axes
            plt.tight_layout(pad=0)

            # Create output subdirectory structure
            relative_path = os.path.relpath(root, input_folder)
            output_subfolder = os.path.join(output_folder, relative_path)
            os.makedirs(output_subfolder, exist_ok=True)

            # Save image
            output_path = os.path.join(output_subfolder, filename.replace('.wav', '.png'))
            plt.savefig(output_path, bbox_inches='tight', pad_inches=0)
            plt.close()
            print(f"Saved: {output_path}")
```
